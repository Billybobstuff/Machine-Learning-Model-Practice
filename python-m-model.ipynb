```python
## Load in Data
%pip install pandas
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/dataprofessor/data/refs/heads/master/delaney_solubility_with_descriptors.csv')
df
```

```python
## **Data Preparation**
## data seperation as X and y
y = df['logS']
y
```

```python
X = df.drop('logS', axis=1)
X
```

```python
## Data splitting
%pip install scikit-learn
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=100)
```

```python
X_train # will have 80% of the data. This is used to build the model
```

```python
X_test # will have 20% of the data. This is used as random data for traing to use.
```

```python
## Model Building
## Linear Regression
## Training the model

from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
## Apply the model to make a prediction
y_lr_train_pred = lr.predict(X_train)
y_lr_test_pred = lr.predict(X_test)
```

```python
y_lr_train_pred
```

```python
y_lr_test_pred
```

```python
## Evaluate Model performance

y_train
```

```python
y_lr_train_pred
```

```python
from sklearn.metrics import mean_squared_error, r2_score

lr_train_mse = mean_squared_error(y_train, y_lr_train_pred)
lr_train_r2 = r2_score(y_train, y_lr_train_pred)

lr_test_mse = mean_squared_error(y_test, y_lr_test_pred)
lr_test_r2 = r2_score(y_test, y_lr_test_pred)
```

```python
print('LR MSE (Train): ', lr_train_mse)
print('LR R2 (Train): ', lr_train_r2)
print('LR MSE (Test): ', lr_test_mse)
print('LR R2 (Test): ', lr_test_r2)
```

```python
lr_results = pd.DataFrame(['Linear regression', lr_train_mse, lr_train_r2, lr_test_mse, lr_test_r2]).transpose()
lr_results.columns = ['Method', 'Training MSE', 'Test MSE', 'Test MSE', 'Test R2']
```

```python
lr_results
```

```python
## Random Forest
## Traning the model

from sklearn.ensemble import RandomForestRegressor # if y is quantitative build regressoion model. If categorical build calsification model

rf = RandomForestRegressor(max_depth=2, random_state=100)
rf.fit(X_train, y_train)


```

```python
y_rf_train_pred = rf.predict(X_train)
y_rf_test_pred = rf.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

rf_train_mse = mean_squared_error(y_train, y_rf_train_pred)
rf_train_r2 = r2_score(y_train, y_rf_train_pred)

rf_test_mse = mean_squared_error(y_test, y_rf_test_pred)
rf_test_r2 = r2_score(y_test, y_rf_test_pred)
```

```python
rf_results = pd.DataFrame(['Random Forest', rf_train_mse, rf_train_r2, rf_test_mse, rf_test_r2]).transpose()
rf_results.columns = ['Method', 'Training MSE', 'Test MSE', 'Test MSE', 'Test R2']
```

```python
## Model Comparison

df_models = pd.concat([lr_results, rf_results], axis=0)
df_models
```

```python
df_models.reset_index(drop=True)
```

```python
## Data Visuallization of prediction results
%pip install matplotlib
%pip install numpy

import numpy as np
import matplotlib.pyplot as plt

z = np.polyfit(y_train, y_lr_train_pred, 1)
p = np.poly1d(z)

plt.figure(figsize=(5, 5))
plt.scatter(x=y_train, y=y_lr_train_pred, c='green', alpha=0.3)

plt.plot(y_train, p(y_train), color='red')
plt.ylabel('Predict LogS')
plt.xlabel('Experimental LogS')
```
